```python
import h5py
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

# Open the MATLAB-generated .mat file
f = h5py.File('model_data_WL.mat', 'r')

# Access the 'model_data_WL' group
model_data_WL = f['model_data_WL']

# Extract the KNN training data
X_knn = np.array(model_data_WL['Training_data']).T  # Transpose if necessary

# Extract the labels
Y = np.array(model_data_WL['label']).ravel()  # Flatten to 1D array

# Extract the number of neighbors for KNN
NumNeighbors = int(np.array(model_data_WL['NumNeighbors_for_knn'])[0][0])

# Extract the decision tree training data
X_tree = np.array(model_data_WL['Training_data_normalized']).T  # Transpose if necessary

# Train the KNN model
knn_model = KNeighborsClassifier(n_neighbors=NumNeighbors，metric='manhattan')
knn_model.fit(X_knn, Y)

# Train the decision tree model
decision_tree_model = DecisionTreeClassifier()
decision_tree_model.fit(X_tree, Y)

# Example: Use the trained models for prediction 
# Load or define your test data
# test_features_knn = np.loadtxt('test_features_knn.csv', delimiter=',')
# test_features_tree = np.loadtxt('test_features_tree.csv', delimiter=',')

# Make predictions with KNN model
# knn_predictions = knn_model.predict(test_features_knn)

# Make predictions with decision tree model
# tree_predictions = decision_tree_model.predict(test_features_tree)

# Print the predictions (uncomment if predictions are made)
# print("KNN Predictions:", knn_predictions)
# print("Decision Tree Predictions:", tree_predictions)
```
